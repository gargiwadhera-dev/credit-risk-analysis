```python
from sklearn.metrics import confusion_matrix

cm = confusion_matrix(y_test, y_pred_balanced)

print(cm)
```

```python
from sklearn.metrics import classification_report

print(classification_report(y_test, y_pred_balanced))
```

```python
from sklearn.metrics import roc_auc_score

y_prob_balanced = balanced_model.predict_proba(X_test_scaled)[:, 1]

auc = roc_auc_score(y_test, y_prob_balanced)

print("ROC-AUC:", auc)
```

```python
threshold = 0.40

y_pred_040 = (y_prob_balanced >= threshold).astype(int)

print(y_pred_040[:20])
```

```python
from sklearn.metrics import confusion_matrix

cm_040 = confusion_matrix(y_test, y_pred_040)

print(cm_040)
```

```python
from sklearn.metrics import classification_report

print(classification_report(y_test, y_pred_040))
```

```python
from sklearn.metrics import precision_score, recall_score, f1_score

thresholds = [0.30, 0.40, 0.50, 0.60, 0.70]

for threshold in thresholds:
    y_pred = (y_prob_balanced >= threshold).astype(int)

    precision = precision_score(y_test, y_pred)
    recall = recall_score(y_test, y_pred)
    f1 = f1_score(y_test, y_pred)

    print(
        "Threshold:", threshold,
        "| Precision:", round(precision, 3),
        "| Recall:", round(recall, 3),
        "| F1:", round(f1, 3)
    )
    
```

```python
import numpy as np
from sklearn.metrics import f1_score

thresholds = np.arange(0.10, 0.91, 0.01)

best_threshold = 0
best_f1 = 0

for threshold in thresholds:
    y_pred = (y_prob_balanced >= threshold).astype(int)
    f1 = f1_score(y_test, y_pred)

    if f1 > best_f1:
        best_f1 = f1
        best_threshold = threshold

print("Best threshold:", round(best_threshold, 2))
print("Best F1-score:", round(best_f1, 3))
```

```python
import numpy as np
from sklearn.metrics import f1_score

thresholds = np.arange(0.10, 0.91, 0.01)

best_threshold_rf = 0
best_f1_rf = 0

for threshold in thresholds:
    y_pred = (y_prob_rf >= threshold).astype(int)
    f1 = f1_score(y_test, y_pred)

    if f1 > best_f1_rf:
        best_f1_rf = f1
        best_threshold_rf = threshold

print("Best Random Forest threshold:", round(best_threshold_rf, 2))
print("Best Random Forest F1-score:", round(best_f1_rf, 3))
```
